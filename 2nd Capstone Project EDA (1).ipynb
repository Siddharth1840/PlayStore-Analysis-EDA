# **Project Summary -**

The Play Store dataset contains 10,841 rows and 13 columns, representing a wide range of mobile applications available on the Google Play Store. Each row corresponds to a unique app, and the dataset includes important attributes such as app name, category, rating, number of reviews, size, installs, type (free or paid), price, content rating, genres, last updated date, current version, and required Android version. This dataset is useful for understanding app performance, user preferences, and market trends in the mobile application ecosystem.

One of the key features of the dataset is the Category column, which groups apps into various segments such as Family, Game, Business, Tools, Lifestyle, and more. Among these, the Family and Game categories dominate, indicating that entertainment and general-purpose apps are the most common on the Play Store. This reflects user demand, as these categories attract a broad audience including children, adults, and casual users.

The Rating column provides insight into user satisfaction, with values typically ranging from 1 to 5. Most apps tend to have ratings between 4.0 and 4.5, suggesting that a large number of apps are generally well-received by users. However, there are missing values in this column, indicating that some apps have not received enough user feedback to generate a rating. Handling these missing values is important during analysis.

The Reviews and Installs columns are strong indicators of app popularity. Apps with higher installs usually have a larger number of reviews, showing a positive correlation between these variables. Install counts vary significantly, ranging from a few installs to over 1 billion installs, highlighting the presence of both small-scale and highly popular apps in the dataset. This variation can be used to segment apps based on popularity.

Another important feature is the Size column, which represents the storage requirement of each app. However, this column contains inconsistent formats such as “Varies with device” and different units like MB and KB. Data cleaning is necessary to standardize this column before performing any meaningful analysis.

The dataset also distinguishes between Free and Paid apps through the Type and Price columns. The majority of apps are free, which aligns with current market trends where developers rely on ads or in-app purchases for revenue. Paid apps are relatively fewer, and their prices vary, though most are reasonably low-cost. This suggests that users prefer free apps, making it a competitive space for developers.

The Content Rating column categorizes apps based on age suitability, such as Everyone, Teen, Mature 17+, etc. Most apps fall under the “Everyone” category, indicating that they are designed for a general audience. This is important for developers targeting a wider user base.

Additionally, the Last Updated, Current Version, and Android Version columns provide technical and maintenance-related information. Frequent updates may indicate active development and better app quality, while compatibility with newer Android versions ensures accessibility to more users.

Overall, this dataset provides valuable insights into app distribution, user behavior, and market trends on the Play Store. Before analysis, significant data cleaning is required due to missing values, inconsistent formats, and potential duplicates. Once cleaned, the dataset can be used for exploratory data analysis (EDA), visualization, and predictive modeling to identify factors that influence app success, user engagement, and ratings.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

df['Rating'].fillna(df['Rating'].mean(), inplace=True)
````

````python
df.dropna(subset=['Type', 'Content Rating', 'Current Ver', 'Android Ver'], inplace=True)
````

````python
df['Size'] = df['Size'].replace('Varies with device', np.nan)

def fix_size(val):
    if pd.isna(val):
        return np.nan
    val = str(val).lower()
    if 'm' in val:
        return float(val.replace('m', ''))
    elif 'k' in val:

        return float(val.replace('k', '')) / 1024
    else:

        try:
            return float(val.replace(',', ''))
        except:
            return np.nan

df['Size'] = df['Size'].apply(fix_size)


mean_size = df['Size'].mean()
df['Size'] = df['Size'].fillna(mean_size)
````

````python

df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')
df['Reviews'] = df['Reviews'].astype('Int64')
````

````python
df['Installs'] = df['Installs'].str.replace('+', '')
df['Installs'] = df['Installs'].str.replace(',', '')
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')
````

````python
df['Price'] = df['Price'].str.replace('$', '')
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
````

````python
df['Last Updated'] = pd.to_datetime(df['Last Updated'])
````

````python
df['Year'] = df['Last Updated'].dt.year
df['Month'] = df['Last Updated'].dt.month_name()
````
